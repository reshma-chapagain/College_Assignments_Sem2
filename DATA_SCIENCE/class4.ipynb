```python
df.groupby(by = ['type1']).agg({
    "attack" : ["mean", "sum" , "min" , "max"],
    "hp" : "sum"
}).reset_index()
```

```python
mew_df = df.groupby(by = ['type1']).agg(
  attack_min = pd.NamedAgg(column= "attack",aggfunc = "min"),
  attack_max = pd.NamedAgg(column = "attack" , aggfunc = "max"),
  attack_avg = pd.NamedAgg(column = "attack" , aggfunc= "mean")
).reset_index()
```

```python
mew_df.to_csv("new data.csv" , index = False)
```
