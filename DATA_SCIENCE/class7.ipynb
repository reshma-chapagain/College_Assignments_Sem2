```python
snapshot_date = df['InvoiceDate'].max()
```

```python
rfm_df = df.groupby(by=['CustomerID']).agg(
    InvoiceDate=('InvoiceDate', lambda x: (snapshot_date - x.max()).days)
)
```

```python
rfm_df.columns = ['CustomerID', 'Recency' , 'Frequency', 'Monetary_Value']
```
