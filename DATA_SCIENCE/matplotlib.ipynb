```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10 , 15))
plt.plot([1,2,3,4,5,6], [1,4,9,16,25,30], color = 'purple' ,label='sample value 1')
plt.plot([1,2,3,4,5,6], [1,7,12,14,29,39], color = 'brown' ,label='sample value 2')
plt.title('Sample Graph')
plt.xlabel('sample x value')
plt.ylabel('sample y value')
plt.legend()
plt.savefig('samplegraph.png')
plt.show()
```
